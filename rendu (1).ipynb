# Affluence et temps d'attente à Walt Disney World

**UE :** Visualisation des données
**Membres :** *DI FEDE Santiago, SELMANE Ilyane, TOUPIN Nicolas*

## 1. Présentation du projet

Walt Disney World (Floride) regroupe quatre parcs à thème : **Magic Kingdom**, **Epcot**, **Hollywood Studios**
et **Animal Kingdom**. Leur fréquentation varie fortement selon la saison, le jour de la semaine, les jours fériés
et les événements. Les files d'attente des attractions varient elles aussi.

Ce projet vise à rendre ces variations **lisibles et explorables** grâce à des visualisations interactives,
afin d'aider à mieux préparer une visite.

**Données :** jeu Kaggle *Walt Disney World Wait Times and Crowd Calendar*, qui couvre la période du 01/01/2015
au 30/09/2026. Le pipeline de préparation (étape 2) est détaillé en annexe, dans `etape2_pipeline.ipynb`.

Deux niveaux de détail sont disponibles :

- **L'affluence** est connue **jour par jour** et pour chaque parc.
- **Les temps d'attente** ne sont connus que **par attraction et par année** : moyenne et pic typique,
  en temps affichés.

## 2. Public cible

| Public | Profil | Besoin principal |
|---|---|---|
| **Visiteurs** (clients) | Familles et voyageurs qui préparent leur séjour | Savoir **quand venir**, **quel parc visiter chaque jour** et **quelles attractions privilégier** |
| **Agences de voyages** | Conseillers qui conçoivent des séjours pour leurs clients | Repérer les **périodes calmes ou chargées**, anticiper l'effet des **événements** et suivre les **tendances** d'une année sur l'autre |

Les deux publics partagent la même question : **à quelle affluence s'attendre ?** Les visiteurs l'utilisent pour
leur propre séjour, les agences pour conseiller plusieurs clients et vendre les bonnes dates.

## 3. Objectifs et tâches d'analyse

Chaque tâche suit la logique **vue d'ensemble puis détail** : du calendrier vers un jour, et d'un parc vers une
attraction.

| N° | Objectif | Tâche d'analyse | Public |
|---|---|---|---|
| T1 | Choisir les dates | Repérer les périodes calmes et chargées selon le mois et la semaine, puis zoomer sur un jour précis | Visiteurs, agences |
| T2 | Affiner le calendrier | Comparer l'affluence selon le jour de la semaine et autour des jours fériés | Visiteurs, agences |
| T3 | Ordonner les parcs | Pour une date donnée, voir quel parc est le plus calme par rapport à sa fréquentation habituelle | Visiteurs |
| T4 | Tenir compte des événements | Mesurer l'affluence pendant les festivals d'Epcot et les soirées spéciales | Agences |
| T5 | Prioriser les attractions | Classer les attractions d'un parc selon leur attente moyenne et leur pic typique | Visiteurs |
| T6 | Suivre les tendances | Suivre l'évolution des attentes et de l'affluence depuis 2015, pour un parc puis pour une attraction | Agences |
| T7 | Observer la Lightning Lane | Comparer l'évolution des attentes des attractions éligibles et non éligibles au Magic Kingdom | Agences |


## 4. Techniques de visualisation envisagées

Chaque membre du groupe met en œuvre une technique différente, sur le même jeu de données.

| Membre | Technique | Données | Vue d'ensemble → détail | Tâches |
|---|---|---|---|---|
| DI FEDE Santiago | Calendrier thermique | Affluence par jour | Grille années × semaines → les 4 parcs le jour cliqué | T1, T2, T3 |
| SELMANE Ilyane | Graphique de rangs (*bump chart*) | Attentes par attraction et par année | Classement des attractions de 2015 à 2026 → détail de l'attraction cliquée | T5, T6, T7 |
| TOUPIN Nicolas | Frise « focus + contexte » | Affluence mensuelle et événements | Frise 2015-2026 → zoom sur la période sélectionnée | T4, T6 |

- **Écartées :** les cartes et les graphes de réseau, faute de données adaptées, ainsi que les histogrammes et camemberts, exclus par les consignes.
- **Outils :** Altair ou Plotly, ou D3.js pour plus de liberté.
- **Cohérence :** les trois visualisations utilisent les mêmes couleurs de parc.
